# 第九週 實作練習

```python
# === 載入資料 ===
import pandas as pd
import numpy as np

bank = pd.read_csv('bank.csv', sep=';')

# === 文字轉數值 === 
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
bank.job=le.fit_transform(bank.job)
bank.marital=le.fit_transform(bank.marital)
bank.education=le.fit_transform(bank.education)
bank.default=le.fit_transform(bank.default)
bank.housing=le.fit_transform(bank.housing)
bank.loan=le.fit_transform(bank.loan)
bank.contact=le.fit_transform(bank.contact)
bank.month=le.fit_transform(bank.month)
bank.poutcome=le.fit_transform(bank.poutcome)
bank.y=le.fit_transform(bank.y)
bank

# === PCA降維 ===
from sklearn.decomposition import PCA
X = bank.drop('y', axis=1)
y = bank['y']
pca = PCA(n_components=8)
newX = pca.fit_transform(X)

# === 分割資料集 ===
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(newX, y, test_size=0.2, random_state=0)


# === BAYES分類器 ===
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
gnb.fit(X_train, y_train)
y_gnb_score = gnb.predict_proba(X_test)
gnb.score(X_test, y_test)

from sklearn.metrics import roc_curve, auc
fpr_gnb, tpr_gnb, thresholds_gnb = roc_curve(y_test, y_gnb_score[:,1])
auc(fpr_gnb, tpr_gnb)


# === 隨機森林 ===
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(n_estimators=20, max_depth=4)
clf.fit(X_train, y_train)
y_clf_score = clf.predict_proba(X_test)

gfpr, gtpr, gthreshold = roc_curve(y_test, y_gnb_score[:,1])
cfpr, ctpr, cthreshold = roc_curve(y_test, y_clf_score[:,1])

roc_auc_gussian = auc(gfpr, gtpr)
roc_auc_clf = auc(cfpr, ctpr)
plt.figure()
lw = 3
plt.plot(gfpr, gtpr, color='darkorange', lw=lw, label='GaussianNB (area = %0.2f)' % roc_auc_gussian)
plt.plot(cfpr, ctpr, color='green', lw=lw, label='RandomForest (area = %0.2f)' % roc_auc_clf)


plt.plot([0, 1], [0, 1], color='navy', lw=lw, linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc="lower right")
plt.show()
```

# 第十週 實作練習

```python
# === A Nural Network ===
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split


df = pd.read_csv('iris.csv')
df.head()

x = df.iloc[:, 0:4]
y = df.iloc[:, 4].values

# 資料前處理
X = MinMaxScaler().fit_transform(x)
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)


# 建立模型
mlp = MLPClassifier(
  activation='relu',          # 活化函數
  alpha=0.0001,               # L2 正則化參數
  batch_size='auto',          # 每次更新所用的樣本數
  beta_1=0.9,                 # 優化器超參數
  beta_2=0.999,               # 優化器超參數
  early_stopping=False,       # 是否提前停止
  epsilon=1e-08,              # 防止除零誤差的小值
  hidden_layer_sizes=(100,),  # 隱藏層結構
  learning_rate='constant',   # 學習率調整方式
  learning_rate_init=0.001,   # 初始學習率
  max_fun=15000,              # 最大函數調用次數
  max_iter=200,               # 最大迭代次數
  momentum=0.9,               # 動量參數
  n_iter_no_change=10,        # 若無改善則停止訓練
  nesterovs_momentum=True,    # 使用 Nesterov 動量
  power_t=0.5,                # 學習率指數
  random_state=None,          # 隨機種子
  shuffle=True,               # 是否在每次迭代時打亂資料
  solver='adam',              # 最佳化演算法
  tol=0.0001,                 # 收斂條件容忍度
  validation_fraction=0.1,    # 驗證集比例
  verbose=False,              # 是否顯示訓練過程
  warm_start=False            # 是否接續上次訓練
)

mlp.fit(x_train, y_train)
mlp.score(x_test, y_test)
```

# 第十一週 實作練習

```python
# === APRIORI 演算法 ===

# 購物籃資料
dataset = [
    ['Milk', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
    ['Dill', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
    ['Milk', 'Apple', 'Kidney Beans', 'Yogurt'],
    ['Milk', 'Unicorn', 'Corn', 'Kidney Beans', 'Yogurt'],
    ['Corn', 'Onion', 'Onion', 'Kidney Beans', 'Ice cream', 'Eggs']
]

from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
te_ary

# 將陣列轉換成 DataFrame
import pandas as pd
df = pd.DataFrame(te_ary, columns=te.columns_)
df

# === 使用 APRIORI 演算法 找出 Frequent Itemsets ===
from mlxtend.frequent_patterns import apriori
frequent_itemsets = apriori(df, min_support=0.6) # 最低支持度 60%，在所有交易中，該產品最少佔60%，才符合frequent itemset

frequent_items = apriori(df, min_support=0.6, use_colnames=True)
frequent_items['length'] = frequent_items['itemsets'].apply(lambda x: len(x))
frequent_items

# 假設只想知道擁有 confidence threshold 90% 以上的 frequent items ,min threshold = 0.7。
from mlxtend.frequent_patterns import association_rules
association_rules(frequent_items, metric="confidence", min_threshold=0.7)

# 篩選更多標準，找出更精準的Association Rules
# 首先要在表中,加一個 column 名叫 “antecedent_len” 來顯示 antecedent 的長度:
frequent_items["antecedent_len"] = frequent_items['antecedents'].apply(lambda x: len(x))
# 建立這個表後,就可在表中指名想 filter 的 antecedent 長度 >=2,以及 confidence 和 lift 的要求:
frequent_items[(frequent_items['antecedent_len'] >= 2) &
               (frequent_items['confidence'] >= 0.75) &
               (frequent_items['lift'] >= 1.2)]
```
