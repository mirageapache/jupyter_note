## 載入資料

```python
import pandas as pd
rawData = pd.read_csv('InterimReport/Urban_Air_Quality_and_Health_Impact.csv')
```

## 填補缺失值

```python
rawData.isnull().sum()
# 含有缺失值的欄位 preciptype(622)、snowdepth(71)、stations(933)、Condition_Code(431)
# 使用眾數值或中位數填補缺失值
rawData['preciptype'] = rawData['preciptype'].fillna(rawData['preciptype'].mode()[0])
rawData['snowdepth'] = rawData['snowdepth'].fillna(rawData['snowdepth'].median())
rawData['stations'] = rawData['stations'].fillna(rawData['stations'].mode()[0])
rawData['Condition_Code'] = rawData['Condition_Code'].fillna(rawData['Condition_Code'].mode()[0])

rawData.info()
```

## 類別型資料轉換

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

rawData['datetime'] = le.fit_transform(rawData['datetime'])
rawData['preciptype'] = le.fit_transform(rawData['preciptype'])
rawData['sunrise'] = le.fit_transform(rawData['sunrise'])
rawData['sunset'] = le.fit_transform(rawData['sunset'])
rawData['conditions'] = le.fit_transform(rawData['conditions'])
rawData['description'] = le.fit_transform(rawData['description'])
rawData['icon'] = le.fit_transform(rawData['icon'])
rawData['stations'] = le.fit_transform(rawData['stations'])
rawData['source'] = le.fit_transform(rawData['source'])
rawData['City'] = le.fit_transform(rawData['City'])
rawData['Season'] = le.fit_transform(rawData['Season'])
rawData['Day_of_Week'] = le.fit_transform(rawData['Day_of_Week'])
rawData['Is_Weekend'] = le.fit_transform(rawData['Is_Weekend'])

rawData
```

## Normalization and Standardization

```python
from sklearn.preprocessing import normalize, scale

normalized_data = normalize(rawData)
print("Normalized Data:\n", normalized_data)

standardized_data = scale(rawData)
print("Standardized Data:\n", standardized_data)
```
